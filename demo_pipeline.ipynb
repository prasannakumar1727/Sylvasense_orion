NDVI → canopy density → change flag → AGB range, on one cached demo AOI.
Runs fully offline on cached arrays — no GEE auth needed to review this notebook.

```python
import numpy as np
import sys; sys.path.append('..')
from backend.app.pipeline.preprocessing import compute_ndvi, vegetation_mask
from backend.app.pipeline.segmentation import classical_fallback
from backend.app.pipeline.change_detection import detect_change
from backend.app.pipeline.biomass import estimate_agb
```

```python
# synthetic demo tile (replace with a real cached GeoTIFF read via rasterio)
rng = np.random.default_rng(7)
nir_before = rng.uniform(0.2, 0.6, (64, 64))
red_before = rng.uniform(0.05, 0.2, (64, 64))
ndvi_before = compute_ndvi(nir_before, red_before)
veg_before = vegetation_mask(ndvi_before)
print(classical_fallback(veg_before))
```

```python
# simulate a second date with some canopy loss
nir_after = nir_before * 0.85
ndvi_after = compute_ndvi(nir_after, red_before)
change = detect_change(ndvi_before, ndvi_after)
print({k: v for k, v in change.items() if k != 'diff_map'})
```

```python
density = classical_fallback(veg_before)['canopy_density_pct']
print(estimate_agb(density))
```
